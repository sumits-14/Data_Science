## **19. Create age groups using Pandas `cut()` such as `16-20`, `21-25`, `26-30`, `31-35`, and `36+`, then calculate the average `Overall` and `Potential` for each age group.**

```python
bins = [15, 20, 25, 30, 35, float('inf')]
label = ['16-20', '21-25', '26-30', '31-35', '36+']
ds['Age_Group'] = pd.cut(
    ds['Age'],
    bins=bins,
    labels=label
)
ds.groupby("Age_Group", observed=True)[['Overall', 'Potential']].mean()
```
