```python
import numpy as np
```

```python
# Create 4x4 array of all True values
np.full((4,4),True)   # with full
#np.ones((4,4),dtype=bool)  # with ones
```

```python
# Create 4x4 array of all False values
np.full((4,4),False) #with full
#np.zeros((4,4),dtype=bool) #with zeros
```

```python
# Create 4x4 array of Digonal elements as 1's and Non diagonal elements as 0's
np.diag(np.ones(4))
```

```python
#Create 4x4 array of Digonal elements as True and Non diagonal elements as False
np.diag(np.ones(4,dtype=bool))
```

```python
# Create 4x4 array of Digonal elements as False and Non diagonal elements as True
~np.diag(np.ones(4,dtype=bool))
```

```python
#Create 2D and 3D arrays of 1's
np.ones((3,4)) #2D of 1's
```

```python
np.ones((2,3,2)) #3D of 1's
```

## **Difference of list and numpy array:**

**List:**

-list is built in datatype

-list are mutable

-list use []

-list support indexing and slicing.

-list can be homogeneous or heterogeneous

**Numpy Array:**

-Numpy is external library

-Numpy array are mutable

-Numpy array use []

-Numpy array support indexing and slicing.

-Numpy array store homogeneous data
