```python
import numpy as np
```

```python
# 1. Create 4x4 array of all True values
s1 = np.ones((4, 4), dtype=int)
print(s1)
```

```python
s2 = np.full((4, 4), 1, dtype=bool)
print(s2)
```

```python
# 2. Create 4x4 array of all False values
s3 = np.full((4, 4), 0, dtype=bool)
print(s3)
```

```python
s31 = np.full((4,4), "False")
print(s31)
```

```python
# 3. Create 4x4 array of Digonal elements as 1's and Non diagonal elements as 0's
s4 = np.eye(4, dtype=int)
print(s4)
```

```python
s5 = np.identity(4, dtype=int)
print(s5)
```

```python
# 4.Create 4x4 array of Digonal elements as True and Non diagonal elements as False
s6 = np.eye(4, dtype=bool)
print(s6)
```

```python
s7 = np.identity(4, dtype=bool)
print(s7)
```

```python
# 5.Create 4x4 array of Digonal elements as False and Non diagonal elements as True
s8 = ~np.eye(4, dtype=bool)
print(s8)
```

```python
s9 = ~np.identity(4, dtype=bool)
print(s9)
```
